```python
import pandas as pd

column_names = [
    "CRIM", "ZN", "INDUS", "CHAS", "NOX",
    "RM", "AGE", "DIS", "RAD", "TAX",
    "PTRATIO", "B", "LSTAT", "MEDV"
]

df = pd.read_csv(
    "4) house Prediction Data Set.csv",
    delim_whitespace=True,
    names=column_names
)

print(df.head())
print(df.dtypes)
```

```python
print(df.describe())
```

```python
print(df.isnull().sum())
```

```python
print("Duplicate Rows:", df.duplicated().sum())
```

```python
corr_matrix = df.corr()

plt.figure(figsize=(12,8))

sns.heatmap(
    corr_matrix,
    annot=True,
    cmap='coolwarm'
)

plt.title("Correlation Heatmap")
plt.show()
```

```python
df.hist(figsize=(15,10))

plt.tight_layout()
plt.show()
```

```python
plt.figure(figsize=(12,6))

sns.boxplot(data=df)

plt.xticks(rotation=90)
plt.title("Boxplot")
plt.show()
```

```python
plt.figure(figsize=(8,6))

sns.scatterplot(x=df["RM"], y=df["MEDV"])

plt.title("RM vs MEDV")
plt.xlabel("Average Rooms (RM)")
plt.ylabel("House Price (MEDV)")

plt.show()
```
